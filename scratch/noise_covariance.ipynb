```python
from sbi_bmode import so_utils
import numpy as np
import matplotlib.pyplot as plt
```

```python
fstrs = ["f030", "f040", "f090", "f150", "f230", "f290"]
lmax = 1000
ells = np.arange(lmax + 1)

fig, ax = plt.subplots(figsize=(7,5))
for fstr in fstrs:
    n_ell = so_utils.get_sat_noise(fstr, "baseline", "pessimistic", lmax)
    ax.plot(ells[2:], n_ell[2:] * ells[2:]*(ells[2:]+1)/(2*np.pi), label=fstr)

ax.set_title("SO SAT noise curves pessimistic baseline")
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlabel(r"$\ell$"); ax.set_ylabel(r"$\ell(\ell+1)N_\ell/2\pi\ [\mu K^2]$")
ax.legend()
plt.show()
```
